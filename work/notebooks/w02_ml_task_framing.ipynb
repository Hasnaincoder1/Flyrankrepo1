## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A single hand-written rule (say, "flag pages older than 90 days with dropping impressions") is exactly the `outputs/model_report.md` baseline already in this repo, and it tops out around Precision@50 ≈ 0.24. The signals that actually separate a genuinely declining page from a normal fluctuation — word count, content type, freshness tier, position tier, search volume, competition level — interact with each other differently per client and per content type, which is too many combinations for one if-statement to hold in its head without either missing real decline or flooding the queue with false alarms. That's the gap a model earns its place in: not because the problem is unknowable, but because the pattern is real and too tangled to write by hand.
