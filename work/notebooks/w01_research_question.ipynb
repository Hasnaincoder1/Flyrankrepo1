## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The code cell below loads `data/raw/content_refresh_anonymized.csv` and checks: how much of the library is actually declining, and how much search visibility is on the line for those pages.


```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print("rows:", df.shape[0], "| clients:", df["client_id"].nunique())

down = df[df["trend_direction"] == "down"]
print(f"pages trending down: {len(down):,} of {len(df):,} ({len(down) / len(df):.1%})")

drop_pct = (down["impressions_prev_30d"] - down["impressions_last_30d"]) / down["impressions_prev_30d"].replace(0, pd.NA)
print(f"median impression drop among down-trend pages: {drop_pct.median():.0%} (prev 30d -> last 30d)")

print(f"total prev-30d impressions sitting on down-trend pages: {down['impressions_prev_30d'].sum():,.0f}")
```
